```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

from xgboost import XGBRegressor

import pickle
```

```python
df = pd.read_csv("demand_forecasting.csv")
```

```python
df
```

```python
df.columns
```

```python
# these i/p features will be used in our ml model
features = [
    "Price",
    "Discount",
    "Inventory Level",
    "Promotion",
    "Competitor Pricing",
    "Category"]
```

```python
features
```

```python
target = "Demand"   # demand is our target var
```

```python
X = df[features].copy()     # creates a copy of the feature dataset
```

```python
X    # displays the i/p features
```

```python
y = df[target]      # y containing the target feature
```

```python
y  # displaying thde target values
```

```python
label_encoders = {}   # creating an empty dictionary

categorical_cols = X.select_dtypes(include="object").columns
# showing categorical cols. in X
```

```python
categorical_cols
```

```python
for col in categorical_cols:
    le = LabelEncoder()                # creating label encoder
    X[col] = le.fit_transform(X[col])  # fit and transform
    # fit --- learns unique categories, # transform --- converts categories into numbers
    label_encoders[col] = le   # storing the encoder into dictionary
```

```python
label_encoders   # displaying the encoders that were created
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)   # 80/20
```

```python
X
```

```python
xgb = XGBRegressor(objective = "reg:squarederror", n_jobs = -1)
```

```python
param_dict = {
    "n_estimators" : [200, 300, 500],
    "max_depth" : [3,4,6,8],
    "learning_rate" : [0.01 , 0.05, 0.1],
    "subsample" : [0.7, 0.8, 1.0],
    "colsample_bytree" : [0.7, 0.8, 1.0],
    "min_child_weight" : [1, 3, 5]
}
```

```python
random_search = RandomizedSearchCV(
    estimator = xgb,
    param_distributions = param_dict,
    n_iter = 25,
    scoring = "neg_mean_absolute_error",
    cv = 3,
    verbose = 1,
    n_jobs = -1
)
```

```python
random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

```python
best_model = random_search.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
```

```python
mean_squared_error(y_test, y_pred)** 0.5
```

```python
best_model.feature_importances_
```

```python
feature_importance = pd.Series(
    best_model.feature_importances_,
    index = X.columns
).sort_values(ascending=False)
```

```python
feature_importance
```

```python
feature_importance.plot(kind="bar",title="Feature Importances")
```
